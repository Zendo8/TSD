```python
import tkinter as tk
from tkinter import messagebox
import time

class BinaryHeap:
    def __init__(self):
        self.heap = []

    def parent(self, i):
        return (i - 1) // 2

    def left_child(self, i):
        return 2 * i + 1

    def right_child(self, i):
        return 2 * i + 2

    def swap(self, i, j):
        self.heap[i], self.heap[j] = self.heap[j], self.heap[i]

    def insert(self, key):
        self.heap.append(key)
        self.heapify_up(len(self.heap) - 1)

    def heapify_up(self, i):
        while i > 0 and self.heap[self.parent(i)] < self.heap[i]:
            self.swap(i, self.parent(i))
            i = self.parent(i)

    def heapify_down(self, i):
        size = len(self.heap)
        largest = i
        l = self.left_child(i)
        r = self.right_child(i)

        if l < size and self.heap[l] > self.heap[largest]:
            largest = l
        if r < size and self.heap[r] > self.heap[largest]:
            largest = r

        if largest != i:
            self.swap(i, largest)
            self.heapify_down(largest)

    def build_heap(self):
        for i in range(len(self.heap) // 2, -1, -1):
            self.heapify_down(i)

    def ascending_traversal_indices(self):
        # Восходящий (Post-order): Лево -> Право -> Корень
        def traverse(i):
            if i < len(self.heap):
                traverse(self.left_child(i))
                traverse(self.right_child(i))
                result.append(i)
        result = []
        traverse(0)
        return result

    def descending_traversal_indices(self):
        # Нисходящий (Pre-order): Корень -> Лево -> Право
        def traverse(i):
            if i < len(self.heap):
                result.append(i)
                traverse(self.left_child(i))
                traverse(self.right_child(i))
        result = []
        traverse(0)
        return result

# Функция для отрисовки дерева
def draw_tree(canvas, heap, x, y, level, i=0, highlighted=None, visited=None):
    if highlighted is None: highlighted = -1
    if visited is None: visited = set()
    
    if i < len(heap):
        # Цвета: текущий - красный, пройденный - розовый, обычный - голубой
        color = "red" if i == highlighted else ("#ffcccc" if i in visited else "lightblue")
        
        canvas.create_oval(x - 20, y - 20, x + 20, y + 20, fill=color, outline="black")
        canvas.create_text(x, y, text=str(heap[i]), fill="black")

        x_offset = 300 / (2 ** level) # Увеличил отступ, чтобы узлы не накладывались

        # Левый ребенок
        l_idx = 2 * i + 1
        if l_idx < len(heap):
            x_left = x - x_offset
            y_left = y + 70
            canvas.create_line(x, y + 20, x_left, y_left - 20, fill="black")
            draw_tree(canvas, heap, x_left, y_left, level + 1, l_idx, highlighted, visited)

        # Правый ребенок
        r_idx = 2 * i + 2
        if r_idx < len(heap):
            x_right = x + x_offset
            y_right = y + 70
            canvas.create_line(x, y + 20, x_right, y_right - 20, fill="black")
            draw_tree(canvas, heap, x_right, y_right, level + 1, r_idx, highlighted, visited)

# Функция для анимации обхода и удаления
def animate_removal(canvas, heap_obj, x_range, y_range, traversal_func):
    to_remove = []
    visited = set()
    
    # Фаза 1: Обход и поиск
    for i in traversal_func():
        visited.add(i)
        if x_range <= heap_obj.heap[i] <= y_range:
            to_remove.append(i)
        
        update_canvas(highlighted=i, visited=visited)
        canvas.update()
        time.sleep(0.6) # Немного ускорил, чтобы не ждать вечность

    # Фаза 2: Удаление
    # Сортируем индексы в обратном порядке, чтобы удаление одного не сместило другие
    for i in sorted(to_remove, reverse=True):
        del heap_obj.heap[i]
        update_canvas()
        canvas.update()
        time.sleep(0.5)

    # Фаза 3: Пересборка кучи
    heap_obj.build_heap()
    update_canvas()
    messagebox.showinfo("Готово", "Лишние вершины удалены, куча перестроена.")

# Функция для обновления отображения дерева
def update_canvas(highlighted=None, visited=None):
    canvas.delete("all")
    # Центрируем дерево на холсте
    width = max(1200, root.winfo_width())
    draw_tree(canvas, heap.heap, 600, 50, 1, highlighted=highlighted, visited=visited)
    # Обновляем область прокрутки
    canvas.config(scrollregion=(0, 0, 1200, 1000))
    result_label.config(text=f"Массив кучи: {heap.heap}")

# Обработка кнопок
def on_insert():
    try:
        value = int(entry_value.get())
        heap.insert(value)
        update_canvas()
        entry_value.delete(0, tk.END)
    except ValueError:
        messagebox.showerror("Ошибка", "Пожалуйста, введите целое число.")

def on_remove():
    if not heap.heap:
        messagebox.showwarning("Внимание", "Куча пуста!")
        return
    try:
        x = int(entry_x.get())
        y = int(entry_y.get())
        if traversal_var.get() == "ascending":
            animate_removal(canvas, heap, x, y, heap.ascending_traversal_indices)
        else:
            animate_removal(canvas, heap, x, y, heap.descending_traversal_indices)
    except ValueError:
        messagebox.showerror("Ошибка", "Пожалуйста, введите корректные значения.")

# Инициализация
heap = BinaryHeap()
root = tk.Tk()
root.title("Двоичная куча - Управление диапазоном")
root.geometry("900x800")

# Верхняя панель управления
controls = tk.Frame(root)
controls.pack(side=tk.TOP, fill=tk.X, padx=10, pady=10)

# Добавление
tk.Label(controls, text="Новое значение:", fg="white").grid(row=0, column=0, padx=5)
entry_value = tk.Entry(controls)
entry_value.grid(row=0, column=1, padx=5)
tk.Button(controls, text="Добавить в кучу", command=on_insert).grid(row=0, column=2, padx=5)

# Диапазон
tk.Label(controls, text="Удалять числа от:", fg="white").grid(row=1, column=0, padx=5, pady=10)
entry_x = tk.Entry(controls, width=10)
entry_x.grid(row=1, column=1, sticky="w", padx=5)

tk.Label(controls, text="до:", fg="white").grid(row=1, column=1)
entry_y = tk.Entry(controls, width=8)
entry_y.grid(row=1, column=1, sticky="e", padx=5)

# Выбор обхода
traversal_var = tk.StringVar(value="ascending")
tk.Radiobutton(controls, text="Восходящий", variable=traversal_var, value="ascending", fg="white").grid(row=1, column=2)
tk.Radiobutton(controls, text="Нисходящий", variable=traversal_var, value="descending", fg="white").grid(row=1, column=3)

tk.Button(controls, text="Удалить диапазон", command=on_remove, bg="#ffaaaa").grid(row=1, column=4, padx=10)

# Область с графом и скроллбарами
tree_frame = tk.Frame(root, bd=2, relief=tk.SUNKEN)
tree_frame.pack(padx=10, pady=10, fill=tk.BOTH, expand=True)

canvas = tk.Canvas(tree_frame, width=800, height=600, bg="white")
v_scroll = tk.Scrollbar(tree_frame, orient=tk.VERTICAL, command=canvas.yview)
h_scroll = tk.Scrollbar(tree_frame, orient=tk.HORIZONTAL, command=canvas.xview)
canvas.configure(yscrollcommand=v_scroll.set, xscrollcommand=h_scroll.set)

v_scroll.pack(side=tk.RIGHT, fill=tk.Y)
h_scroll.pack(side=tk.BOTTOM, fill=tk.X)
canvas.pack(side=tk.LEFT, fill=tk.BOTH, expand=True)

# Нижняя метка
result_label = tk.Label(root, text="Массив кучи: []", font=("Courier", 10), fg="white")
result_label.pack(padx=10, pady=5)

root.mainloop()
```
