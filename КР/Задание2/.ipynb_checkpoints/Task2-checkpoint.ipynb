```python
import tkinter as tk
from tkinter import ttk, messagebox, scrolledtext
import random

def create_hash_table(size):
    return [[] for _ in range(size)]

def hash_function(key, size):
    key_str = str(key)
    if len(key_str) >= 2:
        first_two_digits = key_str[:2]
        return sum(int(digit) for digit in first_two_digits) % size
    else:
        return int(key_str) % size

def insert(hash_table, key, value):
    index = hash_function(key, len(hash_table))
    for kvp in hash_table[index]:
        if kvp[0] == key:
            kvp[1] = value
            return
    hash_table[index].append([key, value])

def search(hash_table, key):
    index = hash_function(key, len(hash_table))
    for kvp in hash_table[index]:
        if kvp[0] == key:
            return kvp[1]
    return None

def delete(hash_table, key):
    index = hash_function(key, len(hash_table))
    for i, kvp in enumerate(hash_table[index]):
        if kvp[0] == key:
            removed_value = kvp[1]
            del hash_table[index][i]
            return removed_value
    return None

def update_hash_table_display():
    hash_table_text.config(state=tk.NORMAL)
    hash_table_text.delete(1.0, tk.END)
    for i, chain in enumerate(hash_table):
        if chain:
            chain_text = f"Индекс {i:02d}: " + " -> ".join(f"[{kvp[0]}: {kvp[1]}]" for kvp in chain)
            hash_table_text.insert(tk.END, chain_text + "\n")
        else:
            hash_table_text.insert(tk.END, f"Индекс {i:02d}: ---\n")
    hash_table_text.config(state=tk.DISABLED)

def on_insert():
    try:
        key_input = entry_key.get().strip()
        if not key_input:
            messagebox.showerror("Ошибка", "Заполните поле 'Ключ'.")
            return
        key = int(key_input)
        value = entry_value.get().strip()
        if not value:
            messagebox.showerror("Ошибка", "Введите значение.")
            return
        insert(hash_table, key, value)
        update_hash_table_display()
        entry_key.delete(0, tk.END)
        entry_value.delete(0, tk.END)
    except ValueError:
        messagebox.showerror("Ошибка", "Некорректный ввод: ключ должен быть целым числом.")

def on_search():
    try:
        key_input = entry_search_key.get().strip()
        if not key_input:
            messagebox.showerror("Ошибка", "Введите ключ для поиска.")
            return
        key = int(key_input)
        result = search(hash_table, key)
        if result:
            messagebox.showinfo("Результат поиска", f"Успех!\nКлюч: {key}\nЗначение: {result}")
        else:
            messagebox.showwarning("Результат поиска", f"Объект с ключом {key} не найден.")
    except ValueError:
        messagebox.showerror("Ошибка", "Ошибка поиска: введите целое число.")

def on_delete():
    try:
        key_input = entry_delete_key.get().strip()
        if not key_input:
            messagebox.showerror("Ошибка", "Введите ключ для удаления.")
            return
        key = int(key_input)
        removed_val = delete(hash_table, key)
        if removed_val:
            messagebox.showinfo("Удаление", f"Ключ: {key}\nЗначение: '{removed_val}'\nСтатус: успешно удален.")
            update_hash_table_display()
            entry_delete_key.delete(0, tk.END)
        else:
            messagebox.showwarning("Удаление", "Ошибка: ключ в таблице отсутствует.")
    except ValueError:
        messagebox.showerror("Ошибка", "Ошибка удаления: введите целое число.")

def on_generate():
    try:
        n = int(entry_generate_count.get())
        for _ in range(n):
            key = random.randint(10, 9999)
            value = f"Value_{key}"
            insert(hash_table, key, value)
        update_hash_table_display()
    except ValueError:
        messagebox.showerror("Ошибка", "Введите целое число в поле количества.")

root = tk.Tk()
root.title("Хеш-таблица (Метод цепочек)")

size = 59
hash_table = create_hash_table(size)

control_frame = tk.Frame(root)
control_frame.pack(padx=20, pady=10, fill="x")

tk.Label(control_frame, text="Ключ:").grid(row=0, column=0, sticky="e")
entry_key = tk.Entry(control_frame)
entry_key.grid(row=0, column=1, padx=5, pady=2)
tk.Label(control_frame, text="Значение:").grid(row=1, column=0, sticky="e")
entry_value = tk.Entry(control_frame)
entry_value.grid(row=1, column=1, padx=5, pady=2)

insert_button = tk.Button(control_frame, text="Добавить элемент", command=on_insert)
insert_button.grid(row=2, column=0, columnspan=2, padx=5, pady=(5, 25), sticky="ew")

tk.Label(control_frame, text="Поиск по ключу:").grid(row=3, column=0, sticky="e")
entry_search_key = tk.Entry(control_frame)
entry_search_key.grid(row=3, column=1, padx=5, pady=2)

search_button = tk.Button(control_frame, text="Найти элемент", command=on_search)
search_button.grid(row=4, column=0, columnspan=2, padx=5, pady=(5, 25), sticky="ew")

tk.Label(control_frame, text="Ключ удаления:").grid(row=5, column=0, sticky="e")
entry_delete_key = tk.Entry(control_frame)
entry_delete_key.grid(row=5, column=1, padx=5, pady=2)

delete_button = tk.Button(control_frame, text="Удалить элемент", command=on_delete)
delete_button.grid(row=6, column=0, columnspan=2, padx=5, pady=(5, 25), sticky="ew")

tk.Label(control_frame, text="Кол-во генераций:").grid(row=7, column=0, sticky="e")
entry_generate_count = tk.Entry(control_frame)
entry_generate_count.grid(row=7, column=1, padx=5, pady=2)

generate_button = tk.Button(control_frame, text="Сгенерировать случайные данные", command=on_generate, bg="#e1e1e1")
generate_button.grid(row=8, column=0, columnspan=2, padx=5, pady=(5, 10), sticky="ew")

hash_table_frame = tk.LabelFrame(root, text=" Визуализация ячеек памяти (m=59) ", padx=10, pady=10)
hash_table_frame.pack(padx=10, pady=10, fill="both", expand=True)

hash_table_text = scrolledtext.ScrolledText(hash_table_frame, width=70, height=15, font=("Consolas", 10), state='disabled')
hash_table_text.pack(fill="both", expand=True)

update_hash_table_display()
root.mainloop()
```
