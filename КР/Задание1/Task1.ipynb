```python
import tkinter as tk
from tkinter import messagebox

def fold_logic(n, folds):
    current_paper = list(range(1, n + 1))
    hidden_layers = []
    
    for x in folds:
        if not current_paper:
            break
        if x <= 0:
            continue
        actual_x = min(x, len(current_paper))
        left_part = current_paper[:actual_x]
        current_paper = current_paper[actual_x:]
        if left_part:
            hidden_layers.insert(0, left_part)
            
    return current_paper, hidden_layers

def on_submit():
    try:
        n_val = entry_n.get().strip()
        f_val = entry_folds.get().strip()
        
        if not n_val or not f_val:
            messagebox.showwarning("Внимание", "Заполните все поля")
            return
            
        n = int(n_val)
        if n <= 0:
            messagebox.showerror("Ошибка", "Длина n должна быть больше 0")
            return

        folds = []
        for val in f_val.replace(' ', '').split(','):
            if val:
                folds.append(int(val))
        
        top, layers = fold_logic(n, folds)
        
        display_text = ""
        display_text += f"ВЕРХНИЙ СЛОЙ:\n{top}\n\n"
        
        for i, layer in enumerate(layers, 1):
            display_text += f"Слой {i+1}:\n{layer}\n\n"
            
        full_result = list(top)
        for l in layers:
            full_result.extend(l)
            
        display_text += f"ИТОГОВЫЙ СПИСОК:\n{full_result}"
        
        result_text.config(state=tk.NORMAL)
        result_text.delete(1.0, tk.END)
        result_text.insert(tk.END, display_text)
        result_text.config(state=tk.DISABLED)
        
    except ValueError:
        messagebox.showerror("Ошибка", "Используйте числа и запятые")

def clear_all():
    entry_n.delete(0, tk.END)
    entry_folds.delete(0, tk.END)
    result_text.config(state=tk.NORMAL)
    result_text.delete(1.0, tk.END)
    result_text.config(state=tk.DISABLED)

root = tk.Tk()
root.title("Симулятор сгибания")
root.geometry("600x700")

DARK_BG = "#2b2b2b"
ENTRY_BG = "#3d3d3d"
BTN_BG = "#444444"
TEXT_WINDOW_BG = "#1e1e1e"
TEXT_FG = "#ffffff"

root.configure(bg=DARK_BG)

input_frame = tk.LabelFrame(root, text=" Параметры ", padx=15, pady=10, bg=DARK_BG, fg=TEXT_FG, font=("Arial", 10, "bold"))
input_frame.pack(padx=20, pady=20, fill="x")

input_frame.columnconfigure(1, weight=1)

label_n = tk.Label(input_frame, text="Длина полоски:", bg=DARK_BG, fg=TEXT_FG, width=15, anchor="w")
label_n.grid(row=0, column=0, pady=5, padx=5, sticky="w")
entry_n = tk.Entry(input_frame, bg=ENTRY_BG, fg=TEXT_FG, insertbackground="white", relief="flat")
entry_n.grid(row=0, column=1, pady=5, padx=5, sticky="ew")

label_folds = tk.Label(input_frame, text="Паста сгибов:", bg=DARK_BG, fg=TEXT_FG, width=15, anchor="w")
label_folds.grid(row=1, column=0, pady=5, padx=5, sticky="w")
entry_folds = tk.Entry(input_frame, bg=ENTRY_BG, fg=TEXT_FG, insertbackground="white", relief="flat")
entry_folds.grid(row=1, column=1, pady=5, padx=5, sticky="ew")

btn_frame = tk.Frame(root, bg=DARK_BG)
btn_frame.pack(pady=5)

submit_btn = tk.Button(btn_frame, text="Запустить сгибание", command=on_submit, bg=BTN_BG, fg="black", 
                       activebackground="#555555", activeforeground="black", relief="flat", width=20, pady=5, font=("Arial", 9, "bold"))
submit_btn.pack(side="left", padx=5)

clear_btn = tk.Button(btn_frame, text="Очистить", command=clear_all, bg=BTN_BG, fg="black", 
                      activebackground="#555555", activeforeground="black", relief="flat", width=10, pady=5, font=("Arial", 9, "bold"))
clear_btn.pack(side="left", padx=5)

result_frame = tk.LabelFrame(root, text=" Визуализация слоев ", padx=15, pady=10, bg=DARK_BG, fg=TEXT_FG, font=("Arial", 10, "bold"))
result_frame.pack(padx=20, pady=10, fill="both", expand=True)

text_container = tk.Frame(result_frame, bg=DARK_BG)
text_container.pack(fill="both", expand=True)

# Оставляем только вертикальный скроллбар
v_scroll = tk.Scrollbar(text_container, orient="vertical")

result_text = tk.Text(text_container, font=("Consolas", 11), state=tk.DISABLED, 
                      bg=TEXT_WINDOW_BG, fg=TEXT_FG, relief="flat", 
                      wrap=tk.WORD, # Теперь текст переносится по словам
                      yscrollcommand=v_scroll.set)

v_scroll.config(command=result_text.yview)

v_scroll.pack(side="right", fill="y")
result_text.pack(side="left", fill="both", expand=True)

root.mainloop()
```
